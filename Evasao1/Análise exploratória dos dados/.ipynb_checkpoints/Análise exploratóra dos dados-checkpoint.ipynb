## TEMPO DE PERMANÊNCIA NO CAMPUS

```python
# separando ano do semestre do fim
base['Ano fim'] = base.iloc[:,3].str.split('.',1)
base['Ano fim'] = base['Ano fim'].str.get(0)
```

```python
# criando variável tempo de permanência no campus
tempo = base[base['Ano fim'] != 'SEM INFORMACAO']
tempo = tempo[tempo['Ano de início'] != 'NAO LEMBRO ']
base['tempo no campus'] = pd.to_numeric(tempo['Ano fim']) - pd.to_numeric(tempo['Ano de início'])
```

```python
base['tempo no campus'].value_counts().index
```

```python
sns.set(rc={'figure.figsize':(11.7,8.27)})
ax = sns.countplot(y = base['tempo no campus'],
              data = base,
              order = base['tempo no campus'].value_counts().index)
width = 0.35
valores = base['tempo no campus'].value_counts().values
ax.bar_label(container=ax.containers[0], labels=valores)
ax.set_ylabel('')
ax.set_xlabel('Alunos')
ax.set_title(base['tempo no campus'].name)
plt.show()
```

*Com essa separação dos anos é possível criar uma nova variável chamada tempo de permanência no campus, com isso é possível calular subtraindo o* ***ANO FIM - ANO INÍCIO*** *o resultado é o tempo de permanência do aluno no campus*

# TABELAS DE CONTIGÊNCIA

*As tabelas de contingência são usadas para registrar observações categóricas de duas ou mais variáveis. No uso desse tipo de tabela é comum se pretender investigar se as variáveis estudadas têm alguma associação com a variável alvo, neste caso a nossa variável alvo é a conclusão ou evasão do aluno.*

```python
def percConvert(ser):
  return ser/float(ser[-1])
```

# Hipóteses

*Para realizar o teste de hipótese será utilizado o SPSS e nele as configurações a seguir: qui-quadrado para identificar se existe significância entre as variáveis selecionadas, para mensurar o nível de força da significância se houver será utilizado o teste de V de Cramer e por fim para identificar a quantidade de chances de um caso ocorrer será utilizado o nível de risco.*

**Range do nível de força da significância V de Cramer**

- 0,00 - 0,10 : Insignificante
- 0,10 - 0,20 : Fraco
- 0,20 - 0,40 : Moderado
- 0,40 - 0,60 : Relativamente forte
- 0,60 - 0,80 : Forte
- 0,80 - 1,00 : Muito forte


### Estudantes que não ingressaram em sua primeira opção de universidade, têm maior probabilidade de evadir-se.

```python
h1 = selecionados.iloc[:,[0,4]]
h1.iloc[:,0] = h1.iloc[:,0].replace(0,'concludente')
h1.iloc[:,0] = h1.iloc[:,0].replace(1,'evadido')
h1.iloc[:,1] = h1.iloc[:,1].replace(0,'nao')
h1.iloc[:,1] = h1.iloc[:,1].replace(1,'sim')
h1.to_csv('h1.csv', sep=';',index=False)
```

**Restultado do teste de hipótese**

- qui-quadrado: 0,039
- V de Cramer:  0,136
- Chances:      1,989 

*Ao realizar o teste de hipótese foi identificado o valor de significância do qui-quadrado, sendo ele 0,039 < 0,05 (limite de erro), apresentando relacionamento entre a escolha de curso com a evasão, porém, para identificar qual a força da significância assintótica foi mensurado nas medidas simétricas o teste de V de Cramer onde apresentou valor de 0,136 que segundo a tabela de nível de força é fraca. Ao realizar o teste da estimativa de risco é possível considerar que um aluno ao escolher um curso CeT como segunda opção pode ocorrer 1,98 chance de evadir.*

### Estudantes que não possuem um conhecimento prévio na área do seu curso, têm uma maior probabilidade de evadir-se.

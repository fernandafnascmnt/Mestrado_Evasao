```python
# importações
import pandas as pd
import numpy as np
import seaborn as sns
import random
import pickle
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, average_precision_score, recall_score, f1_score, cohen_kappa_score, classification_report
from sklearn.linear_model import Perceptron
from sklearn.metrics import plot_confusion_matrix
from sklearn.metrics import confusion_matrix
from math import pi
```

```python
# Carregando a base de dados
base = pd.read_csv('../../../base/stataCompleta-Tratada-Modelo3.csv')

base.describe()
```

```python
base.head(-5)
```

```python
# Normalizando os dados entre 0 e 1
#scaler = StandardScaler()
#base.iloc[:,0:-1] = scaler.fit_transform(base.iloc[:,0:-1])
```

```python
base['desfecho'].value_counts()
```

```python
# Separando concluídos e evadidos
baseC = base[base.desfecho == 0]
baseC.describe()
```

```python
baseC.head()
```

```python
# Separando evadidos
baseE = base[base.desfecho == 1]
baseE.describe()
```

```python
baseE.head()
```

```python
# Selecionando 80% para treinamento da base que contém menor número de amostras
qntRegistros = int(len(baseE.index) * 0.8)
```

```python
# Realizando 50 treinamentos
resultados50 = []
resultados50Testes = []
matrizes50 = []
precisaoTreino0 = []
precisaoTreino1 = []
precisaoTeste0 = []
precisaoTeste1 = []
precisao50Treinamento = []
precisao50Teste = []
matrizesConfusao50Teste = []
desvio50Treinamento = []
desvio50Teste = []
recall50Treino0 = []
recall50Treino1 = []
recall50Teste0 = []
recall50Teste1 = []
recall50Treinamento = []
f1score50Treino0 = []
f1score50Treino1 = []
f1score50Teste0 = []
f1score50Teste1 = []
for i in range(50):
    
    # Selecionando amostras aleatória dos evadidos
    nrRegistrosE = random.sample(range(0, len(baseE.index)), qntRegistros)
    
    # Selecionando amostras aleatória dos concluídos
    nrRegistrosC = random.sample(range(0, len(baseC.index)), qntRegistros)
    
    # Separando entre previsores e classe de cada base
    # Características
    previsoresC = baseC.iloc[nrRegistrosC, 0:-1]
    previsoresE = baseE.iloc[nrRegistrosE, 0:-1]
    df = [previsoresC, previsoresE]
    previsores =  pd.concat(df)
    previsores = previsores.loc[:,:].values

    # Classe
    classeC = baseC.iloc[nrRegistrosC, -1]
    classeE = baseE.iloc[nrRegistrosE, -1]
    dfC = [classeC, classeE]
    classe = pd.concat(dfC)
    classe = classe.values    
    
    # Divisão do k-fold
    kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state = i)
    resultados1 = []
    matrizConfusaoTreino1 = []
    precisaoClasse0Treino = []
    precisaoClasse1Treino = []
    precisaoClasse0Teste = []
    precisaoClasse1Teste = []
    matriz = []
    matrizConfusaoTeste1 = []
    for indice_treinamento, indice_teste in kfold.split(previsores, np.zeros(shape=(classe.shape[0], 1))):
        
        # Carregando o classificador
        classificador = Perceptron()
        
        # Realizando treinamento no Classificador
        classificador.fit(previsores[indice_treinamento], classe[indice_treinamento])
        previsoes = classificador.predict(previsores[indice_teste])
        
        # >>>>>>>>>> APLICANDO AS MÉTRICAS <<<<<<<<<<
        # Acurácia
        precisao = accuracy_score(classe[indice_teste], previsoes)
        resultados1.append(precisao)
        
        # Matriz de confusão validação cruzada
        matriz = confusion_matrix(classe[indice_teste], previsoes)
        matrizConfusaoTreino1.append(matriz)
        
        
    # Acurácia do Treinamento
    resultados1 = np.asarray(resultados1)
    media = resultados1.mean()
    resultados50.append(media)
    
    # Matriz de confusão treinamento
    matrizConfusaoTreino1 = sum(matrizConfusaoTreino1) / len(matrizConfusaoTreino1)
    matrizes50.append(matrizConfusaoTreino1)
    
    # Tranformando a matriz de confusão treinamento em um DataFrame
    dfMatrizTreinamento = pd.DataFrame(matrizConfusaoTreino1)
    
    
    # Precisão Treinamento Classe 0
    precisaoClasse0Treino = dfMatrizTreinamento.iloc[0,0] / (dfMatrizTreinamento.iloc[0,0] + dfMatrizTreinamento.iloc[1,0])
    precisaoTreino0.append(precisaoClasse0Treino)
    
    
    # Precisão Treinamento Classe 1
    precisaoClasse1Treino = dfMatrizTreinamento.iloc[1,1] / (dfMatrizTreinamento.iloc[1,1] + dfMatrizTreinamento.iloc[0,1])
    precisaoTreino1.append(precisaoClasse1Treino)
    
    
    # Recall Treinamento Classe 0
    recallTreino0 = dfMatrizTreinamento.iloc[0,0] / (dfMatrizTreinamento.iloc[0,0] + dfMatrizTreinamento.iloc[0,1])
    recall50Treino0.append(recallTreino0)
    
    # Recall Treinamento Classe 1
    recallTreino1 = dfMatrizTreinamento.iloc[1,1] / (dfMatrizTreinamento.iloc[1,1] + dfMatrizTreinamento.iloc[1,0])
    recall50Treino1.append(recallTreino1)
    
    # Cálculo do desvio Padrão do treinamento
    desvioTreinamento = resultados1.std()
    mediaDesvioTreinamento = desvioTreinamento.mean()
    desvio50Treinamento.append(mediaDesvioTreinamento)
    
    # >>>>>>>>>> F1 Score Treinamento <<<<<<<<<<
    # F1 Score Treinamento Classe 0
    f1scoreTreino0 = 2 * (precisaoClasse0Treino * recallTreino0) / (precisaoClasse0Treino + recallTreino0)
    f1score50Treino0.append(f1scoreTreino0)
    
    # F1 Score Treinamento Classe 1
    f1scoreTreino1 = 2 * (precisaoClasse1Treino * recallTreino1) / (precisaoClasse1Treino + recallTreino1)
    f1score50Treino1.append(f1scoreTreino1)
    
    #>>>>>>>>>> Aplicando teste <<<<<<<<<<
    
    # Removendo amostras do treinamento da base dos concluídos
    baseCteste = baseC.drop(baseC.iloc[nrRegistrosC].index, inplace=False)
    
    # Removendo amostras do treinamento da base dos evadidos
    baseEteste = baseE.drop(baseE.iloc[nrRegistrosE].index, inplace=False)
    
    # Concatenando as bases
    df = [baseCteste, baseEteste]
    baseTeste =  pd.concat(df)
    
    # Separando entre previsores e classe da base de testes
    # Características
    previsoresTeste = baseTeste.iloc[:,0:-1].values
    
    # Classe
    classeTeste = baseTeste.iloc[:,-1].values
    
    # Aplicando o modelo na base de teste
    previsoesTeste = classificador.predict(previsoresTeste)
    
    # Acurácia do teste
    precisaoTeste = accuracy_score(classeTeste, previsoesTeste)
    
    # >>>>>>>>>> Matriz de Confusão Teste <<<<<<<<<<
    matrizConfusaoTeste1 = confusion_matrix(classeTeste, previsoesTeste )
    matrizesConfusao50Teste.append(matrizConfusaoTeste1)
    
    # Tranformando a matriz de confusão teste em um DataFrame
    dfMatrizTeste = pd.DataFrame(matrizConfusaoTeste1)
    
    # Precisão Teste Classe 0
    precisaoClasse0Teste = dfMatrizTeste.iloc[0,0] / (dfMatrizTeste.iloc[0,0] + dfMatrizTeste.iloc[1,0])
    precisaoTeste0.append(precisaoClasse0Teste)
    
    # Precisão Treinamento Classe 1
    precisaoClasse1Teste = dfMatrizTeste.iloc[1,1] / (dfMatrizTeste.iloc[1,1] + dfMatrizTeste.iloc[0,1])
    precisaoTeste1.append(precisaoClasse1Teste)
    
    # >>>>>>>>>> Recall Teste <<<<<<<<<<
    # Recall Teste Classe 0
    recallTeste0 = dfMatrizTeste.iloc[0,0] / (dfMatrizTeste.iloc[0,0] + dfMatrizTeste.iloc[0,1])
    recall50Teste0.append(recallTeste0)
    
    # Recall Teste Classe 1
    recallTeste1 = dfMatrizTeste.iloc[1,1] / (dfMatrizTeste.iloc[1,1] + dfMatrizTeste.iloc[1,0])
    recall50Teste1.append(recallTeste1)
      
    # >>>>>>>>>> F1 Score Teste <<<<<<<<<<
    # F1 Score Teste Classe 0
    f1scoreTeste0 = 2 * (precisaoClasse0Teste * recallTeste0) / (precisaoClasse0Teste + recallTeste0)
    f1score50Teste0.append(f1scoreTeste0)
    
    # F1 Score Teste Classe 1
    f1scoreTeste1 = 2 * (precisaoClasse1Teste * recallTeste1) / (precisaoClasse1Teste + recallTeste1)
    f1score50Teste1.append(f1scoreTeste1)
    
    
    # Salvando resultados testes na variável
    resultados50Testes.append(precisaoTeste)
```

```python
# >>>>>>>>>> Gerando matriz de resultados <<<<<<<<<<
# Convertendo resultados em um array

precisaoTreino0  = np.asarray(precisaoTreino0)
precisaoTreino1 = np.asarray(precisaoTreino1)

precisaoTeste0 = np.asarray(precisaoTeste0)
precisaoTeste1 = np.asarray(precisaoTeste1)

recall50Treino0 = np.asarray(recall50Treino0)
recall50Treino1 = np.asarray(recall50Treino1)

recall50Teste0 = np.asarray(recall50Teste0)
recall50Teste1 = np.asarray(recall50Teste1)

f1score50Treino0 = np.asarray(f1score50Treino0)
f1score50Treino1 = np.asarray(f1score50Treino1)

f1score50Teste0 = np.asarray(f1score50Teste0)
f1score50Teste1 = np.asarray(f1score50Teste1)

desvio50Treinamento = np.asarray(desvio50Treinamento)

resultados50 = np.asarray(resultados50)
resultados50Testes = np.asarray(resultados50Testes)

# Adicionando uma coluna ao lado da outra com o comando zip
dados = list(zip(precisaoTreino0, 
                 precisaoTreino1, 
                 precisaoTeste0, 
                 precisaoTeste1,
                 recall50Treino0,
                 recall50Treino1,
                 recall50Teste0,
                 recall50Teste1,
                 f1score50Treino0,
                 f1score50Treino1,
                 f1score50Teste0,
                 f1score50Teste1,
                 desvio50Treinamento,
                resultados50,
                resultados50Testes))

# Criando um DataFrame e nomeando as colunas
resultados = pd.DataFrame(dados, columns = ['precisaoTreino0', 
                                            'precisaoTreino1', 
                                            'precisaoTeste0', 
                                            'precisaoTeste1',
                                            'recall50Treino0',
                                             'recall50Treino1',
                                             'recall50Teste0',
                                             'recall50Teste1',
                                             'f1score50Treino0',
                                             'f1score50Treino1',
                                             'f1score50Teste0',
                                             'f1score50Teste1',
                                             'desvioTreinamento',
                                            'acuraciaTreino', 
                                            'acuraciaTeste'])


```

```python
# Salvando resultados em um arquivo csv
resultados.to_csv('../resultados/Perceptron/resultadosPerceptronBalanceado.csv', index=False, sep=';')
```
